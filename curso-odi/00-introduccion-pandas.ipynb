```python
import pandas as pd
import numpy as np
```

## 1. Crear un DataFrame simple
Un DataFrame es una tabla con filas y columnas, como en Excel.

```python
# Forma 1: con un diccionario
datos = {
    'nombre': ['Ana', 'Bruno', 'Carlos', 'Diana'],
    'edad': [25, 30, 35, 28],
    'ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Madrid'],
    'salario': [30000, 45000, 50000, 35000]
}

df = pd.DataFrame(datos)
df
```

## 2. Explorar los datos
Preguntas iniciales: ¿cuántas filas y columnas? ¿De qué tipo son? ¿Cuáles son los valores?

```python
# Tamaño
print("Forma (filas, columnas):", df.shape)

# Nombres de columnas
print("\nColumnas:", df.columns.tolist())

# Tipos de dato
print("\nTipos:")
print(df.dtypes)
```

```python
# Información resumida
df.info()
```

```python
# Estadísticas numéricas
df.describe()
```

## 3. Seleccionar datos
Formas de elegir columnas, filas o valores específicos.

```python
# Seleccionar una columna (devuelve una Serie)
print("Una columna (df['edad']):")
print(df['edad'])
print("Tipo:", type(df['edad']))
```

```python
# Seleccionar varias columnas (devuelve DataFrame)
print("Varias columnas (df[['nombre', 'edad']]):")
print(df[['nombre', 'edad']])
```

```python
# Seleccionar filas por posición (iloc = integer location)
print("Fila 0 (primera):")
print(df.iloc[0])

print("\nFilas 0 a 2 (excluye 3):")
print(df.iloc[0:3])
```

```python
# Seleccionar fila por índice (loc)
print("Fila con índice 1:")
print(df.loc[1])

print("\nFila con índice 1, columna 'nombre':")
print(df.loc[1, 'nombre'])
```

## 4. Filtrar (condiciones)
Seleccionar solo las filas que cumplen una condición.

```python
# Personas de 30 años o más
print("Personas con edad >= 30:")
print(df[df['edad'] >= 30])
```

```python
# Personas de Madrid
print("Personas de Madrid:")
print(df[df['ciudad'] == 'Madrid'])
```

```python
# Combinar condiciones
print("Personas de Madrid con edad < 30:")
print(df[(df['ciudad'] == 'Madrid') & (df['edad'] < 30)])
```

## 5. Transformar datos
Crear nuevas columnas o modificar existentes.

```python
# Crear una nueva columna
df['edad_en_5_anos'] = df['edad'] + 5
print(df)
```

```python
# Crear una columna con condición
df['categoria_edad'] = df['edad'].apply(lambda x: 'joven' if x < 30 else 'senior')
print(df[['nombre', 'edad', 'categoria_edad']])
```

```python
# O más sencillo con np.where
df['categoria_edad_v2'] = np.where(df['edad'] < 30, 'joven', 'senior')
print(df[['nombre', 'edad', 'categoria_edad_v2']])
```

## 6. Cálculos y resumen
Agregaciones: suma, promedio, conteo, etc.

```python
# Operaciones simples en una columna
print("Edad promedio:", df['edad'].mean())
print("Edad máxima:", df['edad'].max())
print("Edad mínima:", df['edad'].min())
print("Total de filas:", df['edad'].count())
```

```python
# Agrupar por una columna (groupby)
print("Salario promedio por ciudad:")
print(df.groupby('ciudad')['salario'].mean())
```

```python
# Contar cuántos hay en cada ciudad
print("Personas por ciudad:")
print(df['ciudad'].value_counts())
```

```python
# Múltiples agregaciones
print("Resumen por ciudad:")
print(df.groupby('ciudad')[['edad', 'salario']].agg({'edad': 'mean', 'salario': 'sum'}).round(1))
```

## 7. Datos faltantes (NaN)
Qué hacer cuando faltan valores.

```python
# Crear un DataFrame con valores faltantes
datos2 = {
    'nombre': ['Ana', 'Bruno', 'Carlos', 'Diana'],
    'edad': [25, np.nan, 35, 28],
    'salario': [30000, 45000, np.nan, 35000]
}

df2 = pd.DataFrame(datos2)
print(df2)
```

```python
# Detectar valores faltantes
print("Valores faltantes:")
print(df2.isna())

print("\nCuántos faltantes por columna:")
print(df2.isna().sum())
```

```python
# Eliminar filas con valores faltantes
print("Sin filas faltantes:")
print(df2.dropna())
```

```python
# Llenar valores faltantes
print("Llenados con la media:")
df2_lleno = df2.copy()
df2_lleno['edad'] = df2_lleno['edad'].fillna(df2_lleno['edad'].mean())
df2_lleno['salario'] = df2_lleno['salario'].fillna(df2_lleno['salario'].mean())
print(df2_lleno)
```
